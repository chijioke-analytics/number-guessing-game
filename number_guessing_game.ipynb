```python
import random
numbers = random.randint(1, 100)
attemps = 5

while attemps > 0 :
  user_guess = int(input(f"guess the number 1 to 100 : you have {attemps} attemps :"))
  if user_guess > numbers :
    print("LOWER!")
  elif user_guess < numbers:
    print("HIGHER!")
  else:
    print("Congratulations you have won the game")
    break

  attemps -= 1

  if user_guess != numbers and attemps == 0 :
    print(f"Game over : The number you are trying to guess is {numbers}")
```
